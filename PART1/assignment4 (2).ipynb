```python
import numpy as np

# Function for Union
def fuzzy_union(A, B):
    return np.maximum(A, B)

# Function for Intersection
def fuzzy_intersection(A, B):
    return np.minimum(A, B)

# Function for Complement
def fuzzy_complement(A):
    return 1 - A

# Function for Difference
def fuzzy_difference(A, B):
    return np.minimum(A, 1 - B)

# Cartesian Product
def cartesian_product(A, B):
    return np.outer(A, B)

# Max-Min Composition
def max_min_composition(R, S):
    result = np.zeros((R.shape[0], S.shape[1]))

    for i in range(R.shape[0]):
        for j in range(S.shape[1]):
            result[i, j] = np.max(np.minimum(R[i, :], S[:, j]))

    return result

# Fuzzy Sets
A = np.array([0.2, 0.4, 0.6, 0.8])
B = np.array([0.3, 0.5, 0.7, 0.9])

# Operations
union_result = fuzzy_union(A, B)
intersection_result = fuzzy_intersection(A, B)
complement_result = fuzzy_complement(A)
difference_result = fuzzy_difference(A, B)

# Display Results
print("Union:", union_result)
print("Intersection:", intersection_result)
print("Complement of A:", complement_result)
print("Difference:", difference_result)

# Fuzzy Relations
R = np.array([
    [0.2, 0.5, 0.4],
    [0.6, 0.1, 0.8]
])

S = np.array([
    [0.6, 0.3],
    [0.7, 0.5],
    [0.2, 0.9]
])

# Cartesian Product
cartesian_result = cartesian_product(A, B)

# Max-Min Composition
composition_result = max_min_composition(R, S)

# Display Results
print("\nCartesian Product of A and B:")
print(cartesian_result)

print("\nMax-Min Composition of R and S:")
print(composition_result)
```
